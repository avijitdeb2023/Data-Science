```python
import numpy as np


data = np.array([
    [2, 5, 0],
    [3, 4, 1],
    [4, 6, 0],
    [5, 3, 1],
    [6, 5, 0],
    [7, 4, 1],
    [3, 6, 1],
    [5, 5, 0],
    [8, 3, 1],
    [6, 4, 0],
    [4, 3, 1],
    [7, 6, 0]
], dtype=np.float64)

y = np.array([
    12, 19, 14, 25, 19, 28,
    17, 18, 31, 22, 23, 20
], dtype=np.float64)


# Part (a):

X = np.column_stack((np.ones(len(y)), data))

A = X.T @ X
b = X.T @ y

print("PART (a): DESIGN MATRIX AND NORMAL EQUATIONS\n")


print("Design matrix X:")
print(X.astype(int))

print("\nObservation vector y:")
print(y.astype(int).reshape(-1, 1))

print("\nCoefficient matrix X^T X:")
print(A.astype(int))

print("\nRight-hand side vector X^T y:")
print(b.astype(int).reshape(-1, 1))


# Parts (b) and (c): Explicit Conjugate Gradient implementation


def conjugate_gradient(A, b, tolerance=1e-6, max_iterations=100):
    """
    Solve A @ beta = b using explicit CG iterations.

    Initial guess:
        beta^(0) = [0, 0, 0, 0]^T

    Stopping criterion:
        max(abs(b - A @ beta)) < tolerance

    Returns:
        beta    : final parameter estimates
        history : initial and subsequent iteration results
    """


    beta = np.zeros(b.size, dtype=np.float64)

    r = b - A @ beta
    p = r.copy()
    rr = r @ r

    residual_norm = np.max(np.abs(r))

    history = [(0, beta.copy(), residual_norm)]

    if residual_norm < tolerance:
        return beta, history

    for k in range(1, max_iterations + 1):


        Ap = A @ p

        denominator = p @ Ap

        if not np.isfinite(denominator) or denominator <= 0:
            raise ArithmeticError(
                "CG encountered an invalid p^T A p."
            )

        alpha = rr / denominator



        beta = beta + alpha * p

        r = r - alpha * Ap


        actual_residual = b - A @ beta
        residual_norm = np.max(np.abs(actual_residual))

        history.append((k, beta.copy(), residual_norm))

        if residual_norm < tolerance:
            return beta, history


        rr_new = r @ r

        if not np.isfinite(rr_new) or rr_new <= 0:
            raise ArithmeticError(
                "CG residual breakdown before convergence."
            )

        gamma = rr_new / rr


        p = r + gamma * p


        rr = rr_new

    raise RuntimeError(
        f"CG did not meet the tolerance within {max_iterations} iterations."
    )


beta_hat, history = conjugate_gradient(
    A,
    b,
    tolerance=1e-6,
    max_iterations=100
)

print("\nPARTS (b) AND (c): CONJUGATE GRADIENT ITERATIONS\n")

header = (
    f"{'k':>3}"
    f"{'beta_0':>15}"
    f"{'beta_1':>15}"
    f"{'beta_2':>15}"
    f"{'beta_3':>15}"
    f"{'||r^(k)||_inf':>20}"
)

print(header)
print("-" * len(header))

for k, coefficients, residual_norm in history:
    print(
        f"{k:3d}"
        f"{coefficients[0]:15.6f}"
        f"{coefficients[1]:15.6f}"
        f"{coefficients[2]:15.6f}"
        f"{coefficients[3]:15.6f}"
        f"{residual_norm:20.6e}"
    )

total_iterations = history[-1][0]

print(f"\nTotal number of CG iterations: {total_iterations}")
print(f"Final residual infinity norm: {history[-1][2]:.6e}")
print("Stopping criterion satisfied:",
      history[-1][2] < 1e-6)

print("\nEstimated parameters:")
for j, value in enumerate(beta_hat):
    print(f"beta_{j} = {value:.6f}")

print("\nEstimated regression model:")
print(
    f"y_hat = {beta_hat[0]:.6f}"
    f" {beta_hat[1]:+.6f}*x1"
    f" {beta_hat[2]:+.6f}*x2"
    f" {beta_hat[3]:+.6f}*x3"
)



# Part (d): Predict weekly sales



new_observation = np.array([1, 5, 4, 1], dtype=np.float64)

predicted_sales = new_observation @ beta_hat

print("\nPART (d): PREDICTED WEEKLY SALES\n")
print("Advertising expenditure x1 = 5")
print("Product price x2 = 4")
print("Promotional activity x3 = 1")
print(f"\nPredicted weekly sales = {predicted_sales:.6f}")
```
