```python
# Math 4710 - Homework 3, Question 2(d)
# Successive Over-Relaxation (SOR) method

A = [
    [6.0, -1.0,  2.0],
    [-1.0, 5.0, -2.0],
    [2.0, -2.0,  4.0]
]

b = [7.0, 2.0, 4.0]

TOL = 1e-6
MAX_ITER = 10000


def residual_inf_norm(A, b, x):
    """Compute ||b - Ax||_infinity."""
    n = len(b)

    return max(
        abs(b[i] - sum(A[i][j] * x[j] for j in range(n)))
        for i in range(n)
    )


def sor(A, b, omega, tol=TOL, max_iter=MAX_ITER):
    """Apply SOR from x^(0) = 0 and retain every iteration."""
    n = len(b)
    x = [0.0] * n

    # Include the initial guess, k = 0.
    history = [
        (0, x.copy(), residual_inf_norm(A, b, x))
    ]

    for k in range(1, max_iter + 1):
        x_old = x.copy()

        # Sequential updates: use new values for j < i
        # and previous-iteration values for j > i.
        for i in range(n):
            lower_sum = sum(
                A[i][j] * x[j]
                for j in range(i)
            )

            upper_sum = sum(
                A[i][j] * x_old[j]
                for j in range(i + 1, n)
            )

            x[i] = (
                (1.0 - omega) * x_old[i]
                + (omega / A[i][i])
                * (b[i] - lower_sum - upper_sum)
            )

        # Check the residual only after a complete iteration.
        residual = residual_inf_norm(A, b, x)
        history.append((k, x.copy(), residual))

        if residual < tol:
            return x.copy(), k, residual, history

    raise RuntimeError(
        f"SOR did not converge within {max_iter} iterations "
        f"for omega = {omega}."
    )


def main():
    results = []

    for omega in (0.8, 1.0, 1.2):
        x, iterations, residual, history = sor(A, b, omega)

        print(f"\nSOR with omega = {omega:.1f}")
        print(
            f"{'k':>4}"
            f"{'x1^(k)':>18}"
            f"{'x2^(k)':>18}"
            f"{'x3^(k)':>18}"
            f"{'||r^(k)||_inf':>20}"
        )
        print("-" * 78)

        for k, values, rnorm in history:
            print(
                f"{k:4d}"
                f"{values[0]:18.12f}"
                f"{values[1]:18.12f}"
                f"{values[2]:18.12f}"
                f"{rnorm:20.10e}"
            )

        print("\nFinal approximate solution:")
        for i, value in enumerate(x, start=1):
            print(f"x{i} = {value:.12f}")

        print(f"Total iterations = {iterations}")
        print(f"Final residual infinity norm = {residual:.10e}")

        results.append((omega, iterations, residual))

    print("\nConvergence comparison")
    print(f"{'omega':>8}{'Iterations':>14}{'Final residual':>22}")
    print("-" * 44)

    for omega, iterations, residual in results:
        print(f"{omega:8.1f}{iterations:14d}{residual:22.10e}")

    ranked = sorted(results, key=lambda result: result[1])

    print("\nFastest to slowest for this stopping criterion:")
    print(" -> ".join(
        f"omega = {omega:.1f} ({iterations} iterations)"
        for omega, iterations, _ in ranked
    ))


if __name__ == "__main__":
    main()
```
