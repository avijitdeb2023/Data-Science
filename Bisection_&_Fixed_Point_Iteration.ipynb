# **5(a) Bisection Method**

```python
def f(x):
    return x**3 + x - 4

def bisection(a, b, tol=1e-6, Nmax=100):
    fa = f(a)
    fb = f(b)
    if fa * fb >= 0:
        raise ValueError("f(a) and f(b) must have opposite signs")

    results = []
    n = 0
    while n < Nmax:
        c = (a + b) / 2
        fc = f(c)
        results.append((n, a, b, c, fc))

        # Stopping criterion
        if (b - a) / 2 < tol:
            break

        # Decide which subinterval contains the root
        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc
        n += 1

    return results

# Run the bisection method on f(x) = x^3 + x - 4, starting interval [1, 2]
results = bisection(1, 2, tol=1e-6, Nmax=100)

print(f"{'n':>3} {'a_n':>14} {'b_n':>14} {'c_n':>14} {'f(c_n)':>16}")
for n, a, b, c, fc in results:
    print(f"{n:>3} {a:>14.8f} {b:>14.8f} {c:>14.8f} {fc:>16.10f}")

n_final, a_final, b_final, c_final, fc_final = results[-1]
print(f"\nApproximate root: {c_final:.8f}")
print(f"Number of iterations required: {len(results)}")
```

# **5(b) Fixed Point Iteration**

```python
def g(x):
    return (4 - x) ** (1/3)

def fixed_point(x0, tol=1e-6, Nmax=100):
    results = []
    x_prev = x0
    results.append((0, x_prev, None))   # n = 0: just the initial guess

    n = 1
    while n <= Nmax:
        x_curr = g(x_prev)
        diff = abs(x_curr - x_prev)
        results.append((n, x_curr, diff))

        # Stopping criterion
        if diff < tol:
            break

        x_prev = x_curr
        n += 1

    return results

# Run the fixed-point iteration on g(x) = (4-x)^(1/3), starting from x0 = 1.5
results = fixed_point(1.5, tol=1e-6, Nmax=100)

print(f"{'n':>3} {'x_n':>16} {'|x_n - x_(n-1)|':>20}")
for n, x, diff in results:
    if diff is None:
        print(f"{n:>3} {x:>16.10f} {'---':>20}")
    else:
        print(f"{n:>3} {x:>16.10f} {diff:>20.12f}")

n_final, x_final, diff_final = results[-1]
print(f"\nApproximate root: {x_final:.10f}")
print(f"Number of iterations required: {n_final}")
```
